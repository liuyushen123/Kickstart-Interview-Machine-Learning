# Titanic Exploratory Data Analysis (EDA)
**Goal:** Analyze the survival rates of passengers on the Titanic to identify key demographic and financial factors that influenced survival.

```python
# Import core data science libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set visual style for graphs
sns.set_theme(style="whitegrid")
```

```python
# Load the dataset
titanic = sns.load_dataset("titanic")

# Display the first 5 rows to verify
display(titanic.head())
```

### Initial Data Validation
* The dataset contains 891 rows.
* 'age', 'embark_town', and 'deck' contain missing values.
* 'pclass' and 'sex' should be converted to categorical variables for proper analysis.

```python
# Convert appropriate columns to categorical types
titanic['pclass'] = titanic['pclass'].astype('category')
titanic['sex'] = titanic['sex'].astype('category')

print(titanic.info())
```

```python
# Plot overall survival rate
plt.figure(figsize=(6, 4))
sns.countplot(data=titanic, x='survived', palette='Set2')
plt.title("Titanic Survival Count (0 = Died, 1 = Survived)")
plt.show()
```

```python
# Visualize survival rate by passenger class
plt.figure(figsize=(8, 5))
sns.barplot(data=titanic, x='pclass', y='survived', palette='viridis')
plt.title("Survival Probability by Passenger Class")
plt.ylabel("Probability of Survival")
plt.show()
```
