```python
from nltk.corpus import stopwords

stopword_list = set(stopwords.words("chinese"))
print(stopword_list)
print(stopwords.fileids())

stopword_list.update(['@',"'",'.','"','/','!',',',"'ve","...","n't",'$',"'s"])

print(stopword_list)
```

```python
tokenized_data = []

for each in df.loc[:5,"text"].str.lower():
    tokenized_data.append(nltk.word_tokenize(each))

for each in tokenized_data:
    print(each)
```

```python
data = []

for line in tokenized_data:
    processed_line = []
    for word in line:
        if word not in stopword_list:
            processed_line.append(word)
    data.append(processed_line)

for each in data:
    print(each)
```

```python
strings_for_stemming = df.loc[1:100,"text"].str.lower()
strings_for_stemming
```

```python
from nltk import word_tokenize
words_in_each_string = []

for each_string in strings_for_stemming:
    words_in_each_string.append(word_tokenize(each_string))

for sentence in words_in_each_string:
    print(sentence)
```

```python
original_words_list = []
p_stemmed_list = []
s_stemmed_list = []

for each_list in words_in_each_string:
    print(each_list)
    for word in each_list:
        original_words_list.append(word)
        p_stemmed_list.append(p_stemmer.stem(word))
        s_stemmed_list.append(s_stemmer.stem(word))

```

```python
for i in range(len(original_words_list)):
    print(
        f"{original_words_list[i]:15} | "
        f"{p_stemmed_list[i]:15} | "
        f"{s_stemmed_list[i]:15}"
    )
```

```python
text = df.loc[5, "text"]
print(text)

token_list = nltk.word_tokenize(text)
print(token_list)
```

```python
nltk.download('averaged_perceptron_tagger_eng')

tokens_after_SW = []
for word in token_list:
    if word not in stopword_list:
        tokens_after_SW.append(word)
print(tokens_after_SW)

```

```python
tagged_tokens = nltk.pos_tag(tokens_after_SW)
print(tagged_tokens)

nltk.download('tagsets')
```

```python
def get_pos(tag):
    if tag.startswith("J"):
        return "a"      # adjective
    elif tag.startswith("V"):
        return "v"      # verb
    elif tag.startswith("N"):
        return "n"      # noun
    elif tag.startswith("R"):
        return "r"      # adverb
    else:
        return "n"      # 默认返回 noun
        
        
```

```python
#Lemmatization on a sample string. Stopwords have been removed.
for token, tag in tagged_tokens:
    print("Word:",token,"\nLemmatized Word:",lemmatizer.lemmatize(token.lower(),pos = get_pos(tag)))
    print('***********************')
```
